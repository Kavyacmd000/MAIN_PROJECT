**Numerical Column**

*   product_rating
*   product_rating_count
*   price
*   discount



```python
# Convert numeric columns
df['product_rating'] = pd.to_numeric(df['product_rating'], errors='coerce')
df['product_rating_count'] = pd.to_numeric(df['product_rating_count'], errors='coerce')
df['discount'] = pd.to_numeric(df['discount'].replace('none', np.nan), errors='coerce')

# Convert categorical columns
df['main_category'] = df['main_category'].astype('category')
df['sub_category'] = df['sub_category'].astype('category')
df['currency'] = df['currency'].astype('category')
df['country'] = df['country'].astype('category')
```

```python
df.dtypes
```

```python
df['main_category']
```

```python
df.isnull().sum()
```

```python
df['product_rating'] = df['product_rating'].fillna(df['product_rating'].median())
df['product_rating']
```

```python
df['product_rating_count'] = df['product_rating_count'].fillna(0)
df['product_rating_count']
```

```python
df.isnull().sum()
```

```python
df['price'] = df['price'].fillna(df['price'].median())
df['price']
```

```python
df.dtypes
```

```python
df['discount'] = df['discount'].fillna("0.00")
df['discount']
```

```python
df['discount'].dtype
```

```python
df['discount'] = pd.to_numeric(df['discount'], errors='coerce')
df['discount'].dtype
```

```python
df.isnull().sum()
```

```python
df.head(3)
```

**Categorical Column**

*   sale_tag
*   country
*   currency






```python
df.dtypes
```

```python
df.tail()
```

```python
df['sale_tag']
```

```python
df['sale_tag']=df['sale_tag'].fillna("NONE")
df['sale_tag']
```

```python
df['country'] = df['country'].fillna(df['country'].mode()[0])
```

```python
df['currency'] = df.groupby('country',observed=True)['currency'].transform(
    lambda x: x.fillna(x.mode()[0] if not x.mode().empty else 'Unknown')
)
```

```python
df.isnull().sum()
```

```python
df.tail()
```

```python
df['currency'].unique()
```

```python
conversion_rates = {
    'INR': 1,
    'USD': 83,
    'EUR': 90,
    'GBP': 105,
    'AUD': 55,
    'CAD': 60,
    'SGD': 62,
    'AED': 23,
    'SAR': 22,
    'BHD': 220,
    'SEK': 7.5,
    'CHF': 95,
    'THB': 2.3,
    'CLP': 0.09
}
```

```python
df['price_inr'] = df.apply(
    lambda x: x['price'] * conversion_rates.get(x['currency'], 80),  # default fallback
    axis=1
)
```

```python
df['price_inr']
```
