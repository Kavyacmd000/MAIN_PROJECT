## Stage 2 – Data Cleaning and Pre-processing



```python
df.dtypes
```

Method Used

Used dtypes attribute to inspect column datatypes.

Purpose

To verify whether columns are stored in correct formats.

Reason

Correct datatypes are important for statistical calculations, filtering, grouping, and visualization.

**Numerical Column**

*   product_rating
*   product_rating_count
*   discount



Method Used

Used pd.to_numeric() with errors='coerce'.

Purpose

To convert object datatype columns into numerical datatype.

Reason

Numerical analysis such as mean, median, correlation, and visualizations require numeric datatypes.

**Categorical Column**

*   main_category
*   sub_category
*   country
*   currency






Method Used

Used astype('category').

Purpose

To optimize memory usage and improve categorical analysis.

Reason

Columns like country and category contain repeated values. Category datatype reduces memory usage and improves performance.

```python
# Convert numeric columns
df['product_rating'] = pd.to_numeric(df['product_rating'], errors='coerce')
df['product_rating_count'] = pd.to_numeric(df['product_rating_count'], errors='coerce')
df['discount'] = pd.to_numeric(df['discount'].replace('none', 0), errors='coerce')

# Fill remaining missing discount values
df['discount'] = df['discount'].fillna(0)

# Convert categorical columns
df['main_category'] = df['main_category'].astype('category')
df['sub_category'] = df['sub_category'].astype('category')
df['currency'] = df['currency'].astype('category')
df['country'] = df['country'].astype('category')
```

```python
df.dtypes
```

```python
df['product_rating'] = df['product_rating'].fillna(df['product_rating'].median())
df['product_rating'].head()
```

Method Used

Used median imputation.

Purpose

To replace missing numerical values without removing rows.

Reason

Median is less affected by extreme values and works well for skewed data.

```python
df['product_rating_count'] = df['product_rating_count'].fillna(0)
df['product_rating_count'].head()
```

Method Used

Used constant value imputation.

Purpose

To represent products with no available ratings.

Reason

Missing review counts may indicate absence of reviews.

```python
df['currency'].unique()
```

```python
conversion_rates = {
    'INR': 1,
    'USD': 83,
    'EUR': 90,
    'GBP': 105,
    'AUD': 55,
    'CAD': 60,
    'SGD': 62,
    'AED': 23,
    'SAR': 22,
    'BHD': 220,
    'SEK': 7.5,
    'CHF': 95,
    'THB': 2.3,
    'CLP': 0.09
}
```

# FEATURE 1 — price_inr

```python
df['price_inr'] = df.apply(
    lambda x: x['price'] * conversion_rates.get(x['currency'], 80),  # default fallback
    axis=1
)
```

Method Used

Used apply() with lambda function.

Purpose

To standardize prices into Indian Rupees (INR).

Reason

The dataset contains multiple currencies. Converting all prices into INR enables fair comparison across countries.

## Why apply() was used?

The apply() method was used to perform row-wise currency conversion because each row contains different currency values. It dynamically applies the correct exchange rate based on the currency column.

## Why axis=1 was used?

axis=1 performs row-wise operations and allows access to multiple columns such as price and currency simultaneously.

# Feature 1 - price_inr

```python
df['price_inr'] = df.apply(
    lambda x: x['price'] * conversion_rates.get(x['currency'], 80),
    axis=1
)
```

Convert product prices into Indian Rupees.

# FEATURE 2 — rating_category

Convert ratings into categories.

Purpose - Business segmentation of customer satisfaction.

```python
df['rating_category'] = pd.cut(
    df['product_rating'],
    bins=[0,2,4,5],
    labels=['Low','Medium','High']
)
```

* Purpose -
To classify product ratings into customer satisfaction categories.

* Reason -
Business users can understand grouped rating levels more easily than raw numeric ratings.

* Method Used -
Used `pd.cut()` to divide ratings into Low, Medium, and High categories.

```python
df[['product_rating','rating_category']].head()
```

This engineered feature will be used later in Stage 3 for customer rating analysis.

# FEATURE 3 — price_category

Group products into budget levels.

```python
df['price_category'] = pd.cut(
    df['price_inr'],
    bins=[0,5000,20000,10000000],
    labels=['Budget','Mid-Range','Premium']
)
```

* Purpose -
To classify products into pricing segments.

* Reason -
Grouping products into Budget, Mid-Range, and Premium categories improves business interpretation.

* Method Used -
Used `pd.cut()` to divide prices into pricing ranges.

```python
df[['price_inr','price_category']].head()
```

This engineered feature will be used later in Stage 3 for pricing segmentation analysis.

# FEATURE 4 — has_discount

```python
df['has_discount'] = np.where(
    df['discount'] > 0,
    'Yes',
    'No'
)
```

* Purpose -
To identify whether a product has a discount.

* Reason -
Binary discount classification simplifies discount analysis and visualization.

* Method Used -
Used `np.where()` to classify products as Yes or No based on discount values.

```python
df[['discount','has_discount']].head()
```

This engineered feature will be used later in Stage 3 for pricing segmentation analysis.

# FEATURE 5 — rating_popularity

Combine ratings and review counts.

```python
df['rating_popularity'] = (
    df['product_rating'] *
    df['product_rating_count']
)
# Used in  Top popular products analysis
```

* Purpose -
To measure product popularity by combining customer ratings and rating counts.

* Reason -
A product with high ratings and many reviews is generally more reliable and popular than a product with very few reviews.

* Method Used -
Created a new feature by multiplying `product_rating` with `product_rating_count` to generate a popularity score.

```python
df[['product_rating',
    'product_rating_count',
    'rating_popularity']].head()
```

This engineered feature will be used later in Stage 3 for popular product analysis and visualization.

## 2. Bivariate Analysis

Purpose: Analyze relationship between two variables.

# 2.1 Price vs Rating – Scatterplot

```python
plt.figure(figsize=(8,5))
sns.scatterplot(
    x='product_rating',
    y='price_inr',
    data=df
)
plt.title('Price vs Product Rating')
plt.show()
```

# Interpretation
* Chart Used - `Scatterplot`

* Features Used - `product_rating, price_inr`

* Observation - `No strong linear relationship exists between price and ratings.`

* Insight - `Higher-priced products do not necessarily receive better ratings.`

# 2.2 Average Price by Category – Barplot

```python
category_price = (
    df.groupby('main_category',observed=False)['price_inr']
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

short_labels = [label[:25] for label in category_price.index]

plt.figure(figsize=(14,6))

sns.barplot(
    x=category_price.values,
    y=short_labels
)

plt.title('Top 10 Categories by Average Price')
plt.xlabel('Average Price (INR)')
plt.ylabel('Main Category')

plt.tight_layout()

plt.show()
```

# Interpretation
* Chart Used - `Barplot`

* Features Used - `main_category, price_inr`

* Observation - `Some categories have much higher average prices.`

* Insight - `Product pricing differs significantly across categories.`

# 2.3 Country-wise Average Price

```python
country_price = df.groupby('country',observed=False)['price_inr'].mean().sort_values(ascending=False).head(10)

plt.figure(figsize=(14,6))
plt.tight_layout()
sns.barplot(
    x=country_price.values,
    y=country_price.index
)
plt.title('Top Countries by Average Product Price')
plt.show()
```

# Interpretation
* Chart Used - `Barplot`

* Features Used - `country, price_inr`

* Observation - `Certain countries show higher average product pricing.`

* Insight - `Regional pricing strategies differ across international markets.`
